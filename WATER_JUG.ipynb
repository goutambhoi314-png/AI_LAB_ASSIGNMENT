```python
from collections import deque

# Function to solve water jug problem
def water_jug(capacity1, capacity2, target):
    visited = set()
    queue = deque()

    # Initial state: both jugs are empty
    queue.append((0, 0, []))

    while queue:
        jug1, jug2, path = queue.popleft()

        # Skip if the state has already been visited
        if (jug1, jug2) in visited:
            continue

        visited.add((jug1, jug2))

        # Add current state to path
        new_path = path + [(jug1, jug2)]

        # Check if target is reached
        if jug1 == target or jug2 == target:
            return new_path

        # Generate next possible states
        next_states = [
            # Fill jug1
            (capacity1, jug2),

            # Fill jug2
            (jug1, capacity2),

            # Empty jug1
            (0, jug2),

            # Empty jug2
            (jug1, 0),

            # Pour jug1 -> jug2
            (
                jug1 - min(jug1, capacity2 - jug2),
                jug2 + min(jug1, capacity2 - jug2)
            ),

            # Pour jug2 -> jug1
            (
                jug1 + min(jug2, capacity1 - jug1),
                jug2 - min(jug2, capacity1 - jug1)
            )
        ]

        # Add unvisited states to the queue
        for state in next_states:
            if state not in visited:
                queue.append((state[0], state[1], new_path))

    return None


# Main program
jug1 = int(input("Enter capacity of jug1: "))
jug2 = int(input("Enter capacity of jug2: "))
target = int(input("Enter target volume: "))

solution = water_jug(jug1, jug2, target)
from collections import deque

# Function to solve water jug problem
def water_jug(capacity1, capacity2, target):
    visited = set()
    queue = deque()

    # Initial state: both jugs are empty
    queue.append((0, 0, []))

    while queue:
        jug1, jug2, path = queue.popleft()

        # Skip if the state has already been visited
        if (jug1, jug2) in visited:
            continue

        visited.add((jug1, jug2))

        # Add current state to path
        new_path = path + [(jug1, jug2)]

        # Check if target is reached
        if jug1 == target or jug2 == target:
            return new_path

        # Generate next possible states
        next_states = [
            # Fill jug1
            (capacity1, jug2),

            # Fill jug2
            (jug1, capacity2),

            # Empty jug1
            (0, jug2),

            # Empty jug2
            (jug1, 0),

            # Pour jug1 -> jug2
            (
                jug1 - min(jug1, capacity2 - jug2),
                jug2 + min(jug1, capacity2 - jug2)
            ),

            # Pour jug2 -> jug1
            (
                jug1 + min(jug2, capacity1 - jug1),
                jug2 - min(jug2, capacity1 - jug1)
            )
        ]

        # Add unvisited states to the queue
        for state in next_states:
            if state not in visited:
                queue.append((state[0], state[1], new_path))

    return None


# Main program
jug1 = int(input("Enter capacity of jug1: "))
jug2 = int(input("Enter capacity of jug2: "))
target = int(input("Enter target volume: "))

solution = water_jug(jug1, jug2, target)

if solution is None:
    print("No solution exists.")
else:
    print("Solution found:")
    for step in solution:
        print(step)
if solution is None:
    print("No solution exists.")
else:
    print("Solution found:")
    for step in solution:
        print(step)
```
