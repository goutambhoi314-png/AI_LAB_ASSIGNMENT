```python
def dfs(graph,node,visited):
  if node not in visited:
    print(node,end="")
    visited.add(node)

    for neighbor in graph[node]:
      dfs(graph,neighbor,visited)

graph={}
n=int(input("ENTER THE NUMBER OF VERTIXES:"))
for i in range(n):
  vertex=input(f"enter the vertex{i+1}")
  neighbors=input("enter adjacent vertices {vertex}\n (space-separated):").split()
  graph[vertex]=neighbors
start=input("enter the starting vertex:")

print("\nGraph:",graph)
print("DFS traversal :",end="")
visited=set()
dfs(graph,start,visited)
```
