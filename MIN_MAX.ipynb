```python
#min-max algorithm
def minimax(depth,nodeIndex,isMax,scores,height):
  #Base case: leaf node reached
    if depth==height:
        return scores[nodeIndex - (2**height - 1)] # Adjust index for leaf nodes

    if isMax:
      return max(minimax(depth+1,nodeIndex*2,False,scores,height),minimax(depth+1,nodeIndex*2+1,False,scores,height))
    else:
      return min(minimax(depth+1,nodeIndex*2,True,scores,height),minimax(depth+1,nodeIndex*2+1,True,scores,height))

#Main program
height=3
num_leaf_nodes = 2**height
scores=list(map(int,input(f'Enter {num_leaf_nodes} leaf node values separated by spaces : ').split()))

result=minimax(0,0,True,scores,height)
print('\nthe optimal value is:',result)
```
